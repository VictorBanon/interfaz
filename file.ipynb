```python
import pandas as pd
import json

# Load the CSV
df = pd.read_csv("/home/vic/interfaz/data/taxonomy.csv")

# Relevant taxonomy columns in hierarchical order
columns = ["superkingdom", "phylum", "class", "order", "family", "genus", "species"]

# --- 1. FLAT FORMAT (for dropdown UI) ---
flat_taxonomy = {col: sorted(df[col].dropna().unique().tolist()) for col in columns}
with open("/home/vic/interfaz/data/taxonomy_values.json", "w") as f:
    json.dump(flat_taxonomy, f, indent=2)

# --- 2. NESTED FORMAT with taxon levels included ---
def build_nested_taxonomy(df, columns, level=0):
    if level >= len(columns):
        return []

    current_col = columns[level]
    next_level = level + 1
    nested = []

    for taxon_value, group in df.groupby(current_col):
        node = {
            "name": taxon_value,
            "taxon": current_col
        }
        children = build_nested_taxonomy(group, columns, next_level)
        if children:
            node["children"] = children
        nested.append(node)

    return nested

nested_taxonomy = build_nested_taxonomy(df, columns)

with open("/home/vic/interfaz/data/taxonomy_nested.json", "w") as f:
    json.dump(nested_taxonomy, f, indent=2)

print("✅ Both taxonomy_values.json (flat) and taxonomy_nested.json (with taxon) saved.")
```

```python
import os
import json

def folder_tree_only_dirs(path):
    tree = {}
    for entry in os.listdir(path):
        full_path = os.path.join(path, entry)
        if os.path.isdir(full_path):  # Only include directories
            tree[entry] = folder_tree_only_dirs(full_path)
    return tree


base_path = "/home/vic/interfaz/data/philogenie/"  # replace with the actual folder path
folder_tree = folder_tree_only_dirs(base_path)

# Print it
print(json.dumps(folder_tree, indent=2))
```

```python
import os

def rename_coding_files(directory):
    for root, dirs, files in os.walk(directory):
        for filename in files:
            if "non_cod" in filename:
                old_path = os.path.join(root, filename)
                new_filename = filename.replace("non_cod", "non")
                new_path = os.path.join(root, new_filename)
                os.rename(old_path, new_path)
                print(f"Renamed: {old_path} → {new_path}")

# Example usage
rename_coding_files("/home/vic/interfaz/data")
```
