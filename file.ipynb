```python
import pandas as pd
import json

# Load the CSV
df = pd.read_csv("/home/vic/interfaz/data/taxonomy.csv")

# List of relevant columns
columns = ["superkingdom", "phylum", "class", "order", "family", "genus", "species"]

# Build the dictionary
column_values = {col: df[col].dropna().unique().tolist() for col in columns}

# Save to JSON
with open("/home/vic/interfaz/data/taxonomy_values.json", "w") as f:
    json.dump(column_values, f, indent=2)

print("JSON saved successfully.")
```

```python
import os
import json

def folder_tree_only_dirs(path):
    tree = {}
    for entry in os.listdir(path):
        full_path = os.path.join(path, entry)
        if os.path.isdir(full_path):  # Only include directories
            tree[entry] = folder_tree_only_dirs(full_path)
    return tree


base_path = "/home/vic/interfaz/data/philogenie/"  # replace with the actual folder path
folder_tree = folder_tree_only_dirs(base_path)

# Print it
print(json.dumps(folder_tree, indent=2))
```

```python
import os

def rename_coding_files(directory):
    for root, dirs, files in os.walk(directory):
        for filename in files:
            if "non_cod" in filename:
                old_path = os.path.join(root, filename)
                new_filename = filename.replace("non_cod", "non")
                new_path = os.path.join(root, new_filename)
                os.rename(old_path, new_path)
                print(f"Renamed: {old_path} → {new_path}")

# Example usage
rename_coding_files("/home/vic/interfaz/data")
```
